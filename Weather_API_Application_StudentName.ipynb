## 2. Project Description
This application connects to the Open-Meteo REST API to fetch real-time weather data (temperature, relative humidity, and wind speed) based on user-provided geographic coordinates (latitude and longitude). It also utilizes reverse geocoding via geopy to resolve coordinates into readable city and country names, presenting the information in a clean, user-friendly format with error handling.

```python
# 5. Import Libraries
import requests
from geopy.geocoders import Nominatim

print("Libraries imported successfully.")
```

```python
# 6. API Configuration
BASE_URL = "https://api.open-meteo.com/v1/forecast"
DEFAULT_TIMEOUT = 10

# Initialize reverse geocoder
geolocator = Nominatim(user_agent="weather_app")
```

```python
# 7. API Function
def get_weather(latitude, longitude):
    """Fetches weather metrics from the Open-Meteo API."""
    params = {
        "latitude": latitude,
        "longitude": longitude,
        "current": "temperature_2m,relative_humidity_2m,wind_speed_10m"
    }

    try:
        response = requests.get(BASE_URL, params=params, timeout=DEFAULT_TIMEOUT)
        response.raise_for_status()
        return response.json()

    except requests.exceptions.RequestException as e:
        print("Error connecting to the weather service.")
        print("Error:", e)
        return None

def get_location_name(latitude, longitude):
    """Converts latitude and longitude into a city and country name."""
    try:
        location = geolocator.reverse((latitude, longitude), language="en", timeout=5)
        if location:
            address = location.raw.get("address", {})
            city = address.get("city") or address.get("town") or address.get("village") or address.get("state")
            country = address.get("country", "")

            if city and country:
                return f"{city}, {country}"
            elif country:
                return country
            else:
                return location.address
        return "Unknown Location"
    except Exception:
        return "Unknown Location"
```

```python
# 8. Data Processing
def process_weather_data(data):
    """Extracts required metrics from the API JSON response."""
    if not data or "current" not in data:
        return None

    current = data["current"]
    return {
        "temperature": current.get("temperature_2m", "N/A"),
        "humidity": current.get("relative_humidity_2m", "N/A"),
        "wind_speed": current.get("wind_speed_10m", "N/A")
    }
```

```python
# 9. Data Display
def display_weather(processed_data, location_name, latitude, longitude):
    """Prints weather information in a structured, professional format."""
    if processed_data is None:
        print("\nCould not retrieve weather data due to an error.")
        return

    print("\n================================")
    print("      WEATHER INFORMATION       ")
    print("================================")
    print(f"Location    : {location_name}")
    print(f"Coordinates : {latitude}, {longitude}")
    print("--------------------------------")
    print(f"Temperature : {processed_data['temperature']} °C")
    print(f"Humidity    : {processed_data['humidity']} %")
    print(f"Wind Speed  : {processed_data['wind_speed']} km/h")
    print("================================\n")
```

```python
# 10. User Input & 11. Error Handling (Input Validation)
def get_valid_coordinate(prompt, min_val, max_val):
    """Validates user coordinate input."""
    while True:
        try:
            val = float(input(prompt))
            if min_val <= val <= max_val:
                return val
            else:
                print(f"Invalid input: Value must be between {min_val} and {max_val}.")
        except ValueError:
            print("Invalid input: Please enter a valid floating-point number.")

def run_application():
    print("--- Weather API Application ---")
    latitude = get_valid_coordinate("Enter latitude (-90 to 90): ", -90.0, 90.0)
    longitude = get_valid_coordinate("Enter longitude (-180 to 180): ", -180.0, 180.0)

    location_name = get_location_name(latitude, longitude)
    raw_weather = get_weather(latitude, longitude)
    processed_weather = process_weather_data(raw_weather)

    display_weather(processed_weather, location_name, latitude, longitude)
```

```python
# 12. Testing
# Run the application interactively
run_application()
```
